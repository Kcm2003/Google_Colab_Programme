```python
# Q15. Write a Python program to check whether a number is Even or Odd.

value = int(input("Enter your number : "))

if value > 0:
  if value > 0 and value % 2 == 0:
    print(f"{value} is a Even Number")
  else:
    print(f"{value} is a Odd Number")
else:
  print("Zero or Invalid Number")
```

```python
# Q17. Write a Python program to check whether a given year is a Leap Year or not.

year = int(input("Enter a year: "))

if (year % 400 == 0) or (year % 4 == 0 and year % 100 != 0):
    print(f"{year} is a Leap Year")
else:
    print(f"{year} is Not a Leap Year")
```

```python
# Q18. Write a Python program to find the largest among three numbers.

a1 = int(input("Enter first number: "))
b1 = int(input("Enter second number: "))
c1 = int(input("Enter third number: "))

if a1 >= b1 and a1 >= c1:
    print("Largest number is:", a1)
elif b1 >= a1 and b1 >= c1:
    print("Largest number is:", b1)
else:
    print("Largest number is:", c1)
```

```python
# Q20. Write a Python program to create a Simple Calculator.
# The program should:
# Accept two numbers from the user.
# Accept an operator (+, -, *, /).
# Perform the selected operation and display the result.

num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))
operators = input("Enter operator (+, -, *, /): ")

if operators == "+":
    print("Addition Result =", num1 + num2)

elif operators == "-":
    print("Substraction Result =", num1 - num2)

elif operators == "*":
    print("Multiplication Result =", num1 * num2)

elif operators == "/":
    if num2 != 0:
        print("Division Result =", num1 / num2)
    else:
        print("Division by zero is not allowed.")

else:
    print("Plz Provide any operator.")
```
