**1. Class Method**

```python
# Q1. Create a class Student with a class variable school = "NG High School GKPUR". Write a class method to display the school name.

class Student:
    school = "NG High School GKPUR"

    @classmethod
    def show_school(cls):
        print("School Name:", cls.school)

Student.show_school()

```

```python
# Q2. Create a class Student with a class variable school. Write a class method to change the school name.

class Student:
    school = "NG High School"

    @classmethod
    def change_school(cls, new_school):
        cls.school = new_school

Student.change_school("Changed School :" "BK High School")
print(Student.school)
```

```python
# Q3. Create a class Employee with a class variable company = "Microsoft". Display the company name using a class method.

class Employee:
    company = "Microsoft"

    @classmethod
    def show_company(cls):
        print(cls.company)

Employee.show_company()
```

```python
# Q4. Create a class Bank with a class variable interest_rate = 7.5. Write a class method to update the interest rate.

class Bank:
    interest_rate = 7.5

    @classmethod
    def update_rate(cls, rate):
        cls.interest_rate = rate

Bank.update_rate(8.5)
print(Bank.interest_rate)
```

```python
# Q5. Create a class Car with a class variable brand = "TATA". Change the brand to "Mahindra" using a class method.

class Car:
    brand = "TATA"

    @classmethod
    def change_brand(cls, name):
        cls.brand = name

Car.change_brand("Mahindra")
print(Car.brand)
```

```python
# Q6. Create a class Student with name. Create an object using a class method (from_string).
class Student:
    def __init__(self, name):
        self.name = name

    @classmethod
    def from_string(cls, name):
        return cls(name)

s = Student.from_string("Krushna")
print(s.name)
```

**2. Instance Method**

```python
# Q1. Create a class Student with an instance method to display the student's name.

class Student:
    def __init__(self, name):
        self.name = name

    def show_name(self):
        print("Name:", self.name)

s = Student("Krushna")
s.show_name()
```

```python
# Q2. Create an instance method named calculate_area() that returns the area of the rectangle.

class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def calculate_area(self):
        return self.width * self.height

result = Rectangle(5, 4)
print("Area of the Rectangle is :", result.calculate_area())
```

```python
# Q3. Find the squre of the number.

class Number:
    def __init__(self, num):
        self.num = num

    def square(self):
        print("Square:", self.num ** 2)

n = Number(6)
n.square()
```

```python
# Q4. Create an instance method named drink(amount) that subtracts the given amount of water from the bottle.

class WaterBottle:
    def __init__(self, water):
        self.water = water

    def drink(self, amount):
        self.water -= amount
        print("Water Left:", self.water, "ml")

bottle = WaterBottle(550)
bottle.drink(250)
```

```python
# Q5. Create a class BankAccount with an instance variable balance. Write an instance method deposit() to add money and display the updated balance.

class BankAccount:
    def __init__(self, balance):
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print("Balance:", self.balance)

account = BankAccount(10000)
account.deposit(5000)
```

```python
# Q6. Create a class Calculator with an instance method to add two numbers.

class Calculator:
    def __init__(self, a, b):
        self.a = a
        self.b = b

    def add(self):
        print("Sum:", self.a + self.b)

c = Calculator(21, 56)
c.add()
```

**3. Static Method**

```python
# Q1. Create a class Voting with a static method check_age() to check whether a person is eligible to vote.

class Voting:
    @staticmethod
    def check_age(age):
        if age >= 18:
            print("Eligible to Vote")
        else:
            print("Not Eligible")

Voting.check_age(17)
```

```python
# Q2. Write a programme using static method that cheks the range boundry.

class Thermostat:

    @staticmethod
    def is_safe_temperature(temp):
        return 0 <= temp <= 100

print(Thermostat.is_safe_temperature(134))
print(Thermostat.is_safe_temperature(25))
```

```python
# Q3. File extension extractor

class FileManager:
    @staticmethod
    def get_extension(filename):
        if "." not in filename:
            return "No extension"
        return filename.split(".")[-1]

print(FileManager.get_extension("document.pdf"))
print(FileManager.get_extension("readme.JPEG"))
print(FileManager.get_extension("readme"))
```

```python
# Q4. Calculate the area of circle.

class Circle:
    @staticmethod
    def calculate_area(radius):
        print("Area:", 3.14 * radius * radius)

Circle.calculate_area(5)
```

```python
# Q5. Find the maximum of two numbers.

class Calculator:
    @staticmethod
    def maximum(a, b):
        print(f"Maximum between {a} and {b} is :", max(a, b))

Calculator.maximum(15, 10)
```

```python
# Q6. Check a number is evenn or odd.

class Number:
    @staticmethod
    def check_even_odd(n):
        if n % 2 == 0:
            print("Even")
        else:
            print("Odd")

Number.check_even_odd(21)
```
