```python
# Q30. Create a dictionary where:
# - Key = Student Name
# - Value = Marks
# Display all students who scored more than 80 marks.

dict = {
    "Krushna": 82,
    "Chandra": 76,
    "Maharana": 92,
    "Siva": 81,
    "Subham": 79
}

print("Students who scored more than 80 marks are given below:")

for name, marks in dict.items():
    if marks > 80:
        print(name, "=", marks)
```
