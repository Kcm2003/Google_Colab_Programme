```python
name = "krUsHna ChaNdra"
print(name.upper())
print(name.lower())
print(name.title())
print(name.capitalize())
print(name.swapcase())
print(name.find("Cha"))
print(name.find("test"))
print(name.count("a"))
print(name.startswith("Kr"))
print(name.startswith("Cha"))
print(name.startswith("kr"))
print(name.endswith("Nd"))
print(name.endswith("ra"))

age = "21"
print(name.isdigit())
print(age.isdigit())

edu = "MCA"
print(age.isalpha())
print(edu.isalpha())

alnum = "12abc"
print(age.isalnum())
```

```python
name = "  My name is Krushna   "
print(name)
print(name.strip())
print(name.lstrip())
print(name.rstrip())


print(name.replace("name", "age"))
print(name.split())
print(name.split(","))

name2 = "apple is, a fruit and banana is a fruit, fruit is good for health"
print(name2.split(","))
print(name2.split("is"))

my_name = "My name is Krushna"
print(my_name.title().split())
```

```python
text = "Python"

print(text.center(20))
print(text.center(20, "-"))  # Only working on single character
print(text.ljust(20, "-"))
print(text.rjust(10, "0"))
print(text.zfill(10))
```
